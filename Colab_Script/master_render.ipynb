## Important Configuration Settings Checklist

### Paths
- `PC_TYPE`: Subfolder name for the point cloud dataset
- `GT_FILENAME`: Ground truth `.xyz` filename
- `COLORMAP`: Matplotlib colormap name

### Point Cloud Settings
- `PT_SIZE`: Point radius in rendered output
- `SLICE_XYZ_COLS`: Set `True` if `.xyz` files have extra columns beyond x,y,z

### Object Transforms
- `OBJ_LOCATION`, `OBJ_ROTATION`, `OBJ_SCALE`: Blender space transforms

### Adding New Point Cloud Types
1. Duplicate any config cell (5.x)
2. Update all values for your new point cloud
3. Add the new key to `RENDER_TYPES` in Master Config (Cell 6)

```python
#@title 7. Core Rendering Logic
%%writefile render.py

import os, sys, shutil
import numpy as np
from scipy.spatial import cKDTree

if '/content' not in sys.path:
    sys.path.append('/content')
from config import ALL_CONFIGS, ALL_GT_OVERRIDES, RENDER_TYPES

os.environ['MPLBACKEND'] = 'Agg'
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

import bpy
import blendertoolbox as bt


def setup_env(CFG, src, local, out):
    print("--- Setting up Environment ---")
    os.makedirs(out, exist_ok=True)
    if os.path.exists(local):
        shutil.rmtree(local)
    print("Copying point clouds from Drive to Local...")
    shutil.copytree(src, local)
    print("Copy complete!\n")


def render(CFG, GT_OV, local, out):
    print("--- Starting Render Process ---")
    gt_path = os.path.join(local, CFG["GT_FILENAME"])
    sl = CFG.get("SLICE_XYZ_COLS", False)
    gt_pts = np.loadtxt(gt_path)[:, :3] if sl else np.loadtxt(gt_path)
    tree = cKDTree(gt_pts)

    for fn in os.listdir(local):
        if not fn.endswith(".xyz"): continue
        is_gt = (fn == CFG["GT_FILENAME"])
        ac = CFG.copy()
        if is_gt: ac.update(GT_OV)

        ro = os.path.join(out, f"{os.path.splitext(fn)[0]}.png")
        if os.path.exists(ro) and not ac["FORCE_OVERWRITE"]:
            print(f"Exists: {fn}. Skipping...")
            continue

        print(f"Processing: {fn} {'(GT)' if is_gt else ''}")
        bt.blenderInit(ac["IMG_RES_X"], ac["IMG_RES_Y"], ac["NUM_SAMPLES"], ac["EXPOSURE"])

        if is_gt:
            pts = gt_pts
        elif sl:
            pts = np.loadtxt(os.path.join(local, fn))[:, :3]
        else:
            pts = np.loadtxt(os.path.join(local, fn))

        if is_gt:
            cmap = plt.get_cmap(ac["COLORMAP"])
            colors = np.tile(cmap(0.0)[:3], (pts.shape[0], 1))
        else:
            d, _ = tree.query(pts)
            mx = np.percentile(d, 99)
            nd = np.clip(d / mx, 0, 1)
            nd = np.log1p(nd) / np.log1p(1)
            cmap = plt.get_cmap(ac["COLORMAP"])
            colors = cmap(nd)[:, :3]

        mesh = bt.readNumpyPoints(pts, ac["OBJ_LOCATION"], ac["OBJ_ROTATION"], ac["OBJ_SCALE"])
        mesh = bt.setPointColors(mesh, colors)
        ptC = bt.colorObj(ac["PT_COLOR"], 0.5, 1.0, 1.0, 0.0, 0.0)
        bt.setMat_pointCloudColored(mesh, ptC, ac["PT_SIZE"])

        cam = bt.setCamera(ac["CAM_LOCATION"], ac["LOOK_AT"], ac["FOCAL_LENGTH"])
        sun = bt.setLight_sun(ac["LIGHT_ANGLE"], ac["LIGHT_STRENGTH"], ac["SHADOW_SOFTNESS"])
        bt.setLight_ambient(color=ac["AMBIENT_COLOR"])
        bt.shadowThreshold(alphaThreshold=ac["SHADOW_THRESHOLD"], interpolationMode='CARDINAL')

        bpy.context.scene.use_nodes = True
        tn = bpy.context.scene.node_tree
        tn.nodes.clear()
        rl = tn.nodes.new('CompositorNodeRLayers')
        dn = tn.nodes.new(type='CompositorNodeDenoise')
        co = tn.nodes.new('CompositorNodeComposite')
        vw = tn.nodes.new('CompositorNodeViewer')
        rl.location = (-300, 0)
        dn.location = (0, 0)
        co.location = (300, 0)
        vw.location = (300, -200)
        tn.links.new(rl.outputs['Image'], dn.inputs['Image'])
        tn.links.new(dn.outputs['Image'], co.inputs['Image'])
        tn.links.new(dn.outputs['Image'], vw.inputs['Image'])

        cp = os.path.join(out, f"{ac['COLORMAP']}_colormap.png")
        if not os.path.exists(cp):
            g = np.linspace(0, 1, 256).reshape(1, -1)
            g = np.vstack([g] * 50)
            plt.figure(figsize=(6, 1))
            plt.imshow(g, aspect="auto", cmap=ac["COLORMAP"])
            plt.axis("off")
            plt.savefig(cp, bbox_inches="tight", pad_inches=0, dpi=300)
            plt.close()
            print(f"Colormap saved at {cp}")

        if ac["SAVE_BLEND_FILE"]:
            bp = os.path.join(out, f"{os.path.splitext(fn)[0]}.blend")
            bpy.ops.wm.save_mainfile(filepath=bp)
            print(f"Blend saved at {bp}")

        bt.renderImage(ro, cam)
        print(f"Render complete for {fn}\n")


if __name__ == "__main__":
    for pct in RENDER_TYPES:
        print(f"\n{'='*60}")
        print(f"  RENDERING: {pct}")
        print(f"{'='*60}\n")
        C = ALL_CONFIGS[pct]
        G = ALL_GT_OVERRIDES.get(pct, {})
        src = os.path.join(C["DRIVE_BASE_PATH"], "PointCloud", "xyzFormat", C["PC_TYPE"])
        out = os.path.join(C["DRIVE_BASE_PATH"], "RenderImages", C["PC_TYPE"], f"{C['COLORMAP']}_colormap")
        loc = os.path.join(C["LOCAL_COLAB_BASE"], C["PC_TYPE"])
        setup_env(C, src, loc, out)
        render(C, G, loc, out)
    print("\n\U0001f389 All rendering complete!")
```

```python
#@title 9. Extra: 3D PointCloud Visualizer
import numpy as np
import plotly.graph_objs as go

PC_PATH = "/content/local_data/boy_01_PC_v2/boy01.xyz"

def view_point_cloud(fp, max_pts=100000):
    print(f"Loading {fp}...")
    try: pts = np.loadtxt(fp)
    except Exception as e: print(f"Error: {e}"); return
    if len(pts) > max_pts:
        idx = np.random.choice(len(pts), max_pts, replace=False)
        pts = pts[idx]
    x,y,z = pts[:,0], pts[:,1], pts[:,2]
    fig = go.Figure(data=[go.Scatter3d(x=x,y=y,z=z,mode='markers',
        marker=dict(size=1.5,color=z,colorscale='Viridis',opacity=0.8))])
    fig.update_layout(scene=dict(aspectmode='data'),
        margin=dict(l=0,r=0,b=0,t=0),height=700)
    fig.show()

view_point_cloud(PC_PATH)
```
